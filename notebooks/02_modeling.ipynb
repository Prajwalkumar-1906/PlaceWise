### 1. Load Evaluation Metrics from Registry

```
with open('reports/metrics.json', 'r') as f:
    metrics = json.load(f)

print('Test Set Performance Comparison:')
summary = pd.DataFrame({
    'Model': ['Logistic Regression', 'Random Forest', 'Tuned XGBoost'],
    'ROC-AUC': [
        metrics['test']['logistic_regression']['roc_auc'],
        metrics['test']['random_forest']['roc_auc'],
        metrics['test']['xgboost']['roc_auc'],
    ],
    'F1 Score': [
        metrics['test']['logistic_regression']['f1'],
        metrics['test']['random_forest']['f1'],
        metrics['test']['xgboost']['f1'],
    ],
    'Brier Score (Calibration)': [
        metrics['test']['logistic_regression']['brier_score'],
        metrics['test']['random_forest']['brier_score'],
        metrics['test']['xgboost']['brier_score'],
    ]
})
display(summary.round(4))
```
